Cell 2: Simulate Traffic Using NumPy

```python

import numpy as np

np.random.seed(42)

# Simulation settings
days = 30
junctions = 10
hours = 24

hour = np.arange(hours)
base_traffic = 20

# Morning and evening traffic peaks
morning_peak = 80 * np.exp(-0.5 * ((hour - 8) / 2) ** 2)
evening_peak = 100 * np.exp(-0.5 * ((hour - 18) / 2.5) ** 2)

# Hourly traffic pattern
hourly_traffic = base_traffic + morning_peak + evening_peak

# Traffic factors for 10 junctions
junction_factors = np.random.uniform(0.5, 1.5, junctions)

# Calculate traffic rates
traffic_rates = (
    hourly_traffic[None, None, :]
    * junction_factors[None, :, None]
)

# Generate traffic data for 30 days
traffic_data = np.random.poisson(
    lam=np.broadcast_to(
        traffic_rates,
        (days, junctions, hours)
    )
)

print("Traffic Simulation Completed!")
print("Dataset Shape:", traffic_data.shape)
print("Total Records:", traffic_data.size)

print("\nTraffic for Day 1, Junction 1:")
print(traffic_data[0, 0, :])
```

Cell 3: Calculate Total Traffic and Find Peak Hours


```python

# Calculate total traffic at each junction
junction_totals = np.sum(traffic_data, axis=(0, 2))

# Calculate total traffic for each hour
hourly_totals = np.sum(traffic_data, axis=(0, 1))

# Calculate average traffic for each hour
hourly_averages = np.mean(traffic_data, axis=(0, 1))

# Find the busiest hour
peak_hour = np.argmax(hourly_averages)

# Find the busiest junction
busiest_junction = np.argmax(junction_totals)

print("Total Traffic at Each Junction:")

for i in range(junctions):
    print("Junction", i + 1, ":", junction_totals[i])

print("\nBusiest Hour:", peak_hour, ":00")
print("Average Traffic:", hourly_averages[peak_hour])

print("\nBusiest Junction:", busiest_junction + 1)
print("Total Vehicles:", junction_totals[busiest_junction])
```

Cell 4: Traffic Heatmap Data and Performance Comparison
import time



```python

import time

# Calculate average traffic for each junction and hour
junction_hour_average = np.mean(traffic_data, axis=0)

print("Average Traffic for Each Junction and Hour:")
print(junction_hour_average)

# Find the busiest junction and hour
busiest_position = np.argmax(junction_hour_average)

busiest_junction, busiest_hour = np.unravel_index(
    busiest_position, junction_hour_average.shape
)

print("\nBusiest Junction:", busiest_junction + 1)
print("Busiest Hour:", busiest_hour, ":00")

# Loop-based calculation
start = time.time()

loop_totals = np.zeros((junctions, hours))

for d in range(days):
    for j in range(junctions):
        for h in range(hours):
            loop_totals[j, h] += traffic_data[d, j, h]

loop_time = time.time() - start

# NumPy vectorized calculation
start = time.time()

vectorized_totals = np.sum(traffic_data, axis=0)

vectorized_time = time.time() - start

print("\nPerformance Comparison")
print("Loop Execution Time:", loop_time, "seconds")
print("NumPy Execution Time:", vectorized_time, "seconds")

print(
    "Results Match:",
    np.allclose(loop_totals, vectorized_totals)
)
```
